```python
import numpy as np
%pip install pandas

import pandas as pd
import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense

# 1. Loading Dataset
df = pd.read_csv('ANN_health_dataset.csv')


# 2. Prepare Input and Output
X = df[['Diet_Score', 'Exercise_Hours',
        'Sleep_Hours', 'Stress_Level']].values
y = df['Weight_Lost'].values

#3.Build ANN
model = Sequential([
    tf.keras.Input(shape=(4,)),
    Dense(8, activation='relu'),
    Dense(8, activation='relu'),
    Dense(4, activation='relu'),
    Dense(4, activation='relu'),
    Dense(1, activation='sigmoid')
])


# 4. Compile Model
model.compile(
    optimizer='adam',
    loss='binary_crossentropy',
    metrics=['accuracy']
)

model.fit(X, y, epochs=200, verbose=0)

loss, accuracy = model.evaluate(X, y, verbose=1)
print("Accuracy",format(accuracy * 100))

predictions = model.predict(X, verbose=0)

df['Predictions'] = (predictions >= 0.5).astype(int)

print("\n--- Enter Lifestyle Details ---")

diet = float(input("Diet Score (1-10): "))
exc = float(input("Exercise Hours: "))
sleep = float(input("Sleep Hours: "))
stress = float(input("Stress Level (1-10): "))

user_data = np.array([[diet, exc, sleep, stress]])

prediction = model.predict(user_data, verbose=0)[0][0]

#print("\nPrediction:")

if prediction >= 0.5:
    print("Likely to Achieve Weight Loss")
else:
    print("Unlikely to Achieve Weight Loss")



    
```
